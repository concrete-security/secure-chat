# Taux de success par system prompt

```python
import pandas as pd

df = pd.read_csv('../experiments/system_prompt_test/system_prompt_results.csv')

stats = df.groupby(['prompt_name', 'sdk'])['success'].agg(['sum', 'count'])
stats.columns = ['nb_success', 'nb_runs']
stats['success_rate (%)'] = (stats['nb_success'] / stats['nb_runs'] * 100).round(1)
stats = stats.sort_values('success_rate (%)', ascending=False)
stats = stats.astype({'nb_success': int, 'nb_runs': int})
stats


# Le prompt par défaut avant nos modifs était beast (11 079 chars). Et
#   dans le classement actuel, beast est à 58.6% avec anthropic SDK.

```
